```python
from ultralytics import RTDETR
from ultralytics import RTDETR
import torch
from pathlib import Path
import pandas as pd
```

```python
model = RTDETR("rtdetr-l.pt")
print("Loaded successfully")
```

```python
print("Torch:", torch.__version__)

print("CUDA:", torch.cuda.is_available())

print("GPU:", torch.cuda.get_device_name(0))
```

```python
DATA_YAML = "/data/UG/Kiranmoy/datasets/MHCD2022_YOLO/data.yaml"

PROJECT_NAME = "MHCD_Benchmarks"
RUN_NAME = "RTDETR_L_Baseline"
```

```python
results = model.train(
    data=DATA_YAML,
    epochs=100,
    imgsz=640,
    batch=32,
    device=0,
    workers=8,
    seed=42,

    project="runs/detect/MHCD_Benchmarks",
    name="RTDETR_L_Baseline",

    pretrained=True,
    patience=20,
    verbose=True
)
```

```python
print(results.save_dir)
```

```python
best_model_path = Path(results.save_dir) / "weights" / "best.pt"

print(best_model_path)
print(best_model_path.exists())
```

```python
best_model = RTDETR(str(best_model_path))

metrics = best_model.val(
    data=DATA_YAML,
    split="test",
    device=0,
    workers=0
)
```

```python
print("AP50      :", metrics.box.map50)
print("AP75      :", metrics.box.map75)
print("AP50-95   :", metrics.box.map)

print("Precision :", metrics.box.mp)
print("Recall    :", metrics.box.mr)

f1 = (
    2 * metrics.box.mp * metrics.box.mr
    / (metrics.box.mp + metrics.box.mr + 1e-8)
)

print("F1        :", f1)
```
